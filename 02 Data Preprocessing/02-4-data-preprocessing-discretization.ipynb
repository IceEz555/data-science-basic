```python
import pandas as pd
import numpy as np

golf_dataset = pd.read_csv("./data/golf_dataset_mini_allnumerical.csv")
df = golf_dataset.copy()

df
```

# Binning Strategies
![Imgur](https://i.imgur.com/lZmWGDl.png)

# ✤ Equal Width Binning

![Imgur](https://i.imgur.com/U6qXAci.png)

```python
# Equal-width Binning on Temperature
from sklearn.preprocessing import KBinsDiscretizer
temp_discretizer = KBinsDiscretizer(n_bins=4, encode='ordinal', strategy='uniform', subsample=None)
df['Temperature_Binned'] = temp_discretizer.fit_transform(df[['Temperature']]).astype(int)
df[['Temperature','Temperature_Binned']]
```

# ✤ Equal Frequency Binning

![Imgur](https://i.imgur.com/xRFv9ua.png)

```python
# Equal-frequency Binning on Crowdedness
crowdedness_discretizer = KBinsDiscretizer(n_bins=4, encode='ordinal', strategy='quantile', subsample=None)
df['Crowdedness_Binned'] = crowdedness_discretizer.fit_transform(df[['Crowdedness']]).astype(int)
df[['Crowdedness', 'Crowdedness_Binned']]
```

# ✤ Custom Width Binning

![Imgur](https://i.imgur.com/fwmCRza.png)

```python
# Custom Binning on Weekday (binary case)
df['Weekday_Bins'] = np.where(df['Weekday'] <= 4,0,1)
df[['Weekday','Weekday_Bins']]
```

```python
# Custom Binning on Month (multi-categories case)
bins = [0, 3, 6, 9 , 12]
labels = ['Q1', 'Q2', 'Q3', 'Q4']
df['Month_Bins'] = pd.cut(df['Month'], bins=bins, labels=labels)
df[['Month','Month_Bins']]
```

# ✤ Clustering for Binning

![Imgur](https://i.imgur.com/0Twzpwy.png)

```python
# k-means binning on Temperature & Humidity
temp_discretizer = KBinsDiscretizer(n_bins=4, encode='ordinal', strategy='kmeans', subsample=None)
df['Humidity_Binned'] = temp_discretizer.fit_transform(df[['Humidity']]).astype(int)
df[['Humidity','Humidity_Binned']]
```
