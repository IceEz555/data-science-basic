## 1. บทนำและสถาปัตยกรรมของ Google Colab

Google Colaboratory (เรียกโดยย่อว่า Google Colab) เป็นแพลตฟอร์ม Interactive Computing บนระบบคลาวด์ของ Google ซึ่งพัฒนาต่อยอดมาจากโครงการโอเพนซอร์ส Jupyter Notebook

สำหรับผู้บริหารระดับต้นและผู้จัดการ การทำความเข้าใจ Google Colab มีความสำคัญเชิงกลยุทธ์ ดังนี้:
- **การลดต้นทุนด้านโครงสร้างพื้นฐาน (Infrastructure Cost Reduction):** ทีมงานสามารถเริ่มพัฒนาโมเดลหรือวิเคราะห์ข้อมูลได้ทันทีโดยไม่ต้องจัดซื้อเครื่องคอมพิวเตอร์สมรรถนะสูงประจำตำแหน่ง
- **การทำงานร่วมกันแบบเรียลไทม์ (Collaborative Workflow):** สามารถแชร์สมุดงาน (Notebook) พร้อมสิทธิ์การอ่านและการแก้ไขได้เช่นเดียวกับ Google Docs ช่วยให้ผู้บริหารสามารถตรวจสอบตรรกะทางธุรกิจที่ทีมข้อมูลนำไปปฏิบัติได้โดยตรง
- **การผสานเอกสารและผลลัพธ์เชิงประจักษ์ (Executable Documentation):** รวมคำอธิบายบริบททางธุรกิจ โค้ดประมวลผล และกราฟสรุปผลไว้ในเอกสารฉบับเดียวกัน ทำให้การตรวจสอบย้อนกลับ (Auditability) เป็นไปอย่างโปร่งใส

## 2. โครงสร้างของสมุดงาน: Code Cell เทียบกับ Text (Markdown) Cell

เอกสารของ Google Colab ประกอบด้วยหน่วยย่อยที่เรียกว่า "เซลล์ (Cell)" ซึ่งแบ่งออกเป็น 2 ประเภทหลัก:

| ประเภทเซลล์ | หน้าที่หลัก | วัตถุประสงค์สำหรับผู้บริหารและทีมงาน |
|---|---|---|
| **Text Cell (Markdown)** | ใช้บันทึกข้อความ ตาราง หัวข้อ และคำอธิบาย โดยใช้ภาษา Markdown | ใช้ระบุโจทย์ทางธุรกิจ สมมติฐาน ข้อจำกัด และข้อสรุปจากการวิเคราะห์ |
| **Code Cell** | ใช้เขียนคำสั่งภาษา Python เพื่อประมวลผล ดึงข้อมูล หรือคำนวณทางสถิติ | ใช้ตรวจสอบตรรกะการคำนวณของทีมข้อมูล และสั่งประมวลผลเพื่อดูผลลัพธ์จริง |

> **ข้อควรระวังสำหรับผู้บริหาร:** การประมวลผลใน Notebook เป็นไปตามลำดับการสั่งทำงาน (Execution Order) ไม่ได้ขึ้นอยู่กับตำแหน่งบรรทัดบนหน้าจอเพียงอย่างเดียว หมายเลขในวงเล็บ `[ ]` หน้า Code Cell แสดงลำดับการทำงานจริง หากลำดับข้ามไปมา อาจทำให้ผลลัพธ์คลาดเคลื่อนได้

## 3. การจัดการสภาพแวดล้อม: Runtime และ Session Storage

เมื่อผู้ใช้งานเปิดสมุดงานและเริ่มประมวลผล Google จะจัดสรรเครื่องเสมือน (Virtual Machine: VM) บนศูนย์ข้อมูลของ Google ขึ้นมาทำงานเบื้องหลัง ซึ่งเรียกว่า **Runtime**

### 3.1 ทรัพยากรของ Runtime
- **หน่วยประมวลผล (Compute):** CPU หรือตัวเร่งความเร็วฮาร์ดแวร์ เช่น GPU/TPU
- **หน่วยความจำหลัก (RAM):** พื้นที่สำหรับเก็บตัวแปรและตารางข้อมูลขณะกำลังประมวลผล
- **พื้นที่จัดเก็บข้อมูล (Disk):** พื้นที่ดิสก์ของเครื่องเสมือนสำหรับติดตั้งโปรแกรมและวางไฟล์

### 3.2 ความเสี่ยงของ Session Storage ต่อการปฏิบัติงาน
พื้นที่จัดเก็บไฟล์บนเครื่องเสมือนนี้เรียกว่า **Session Storage (พื้นที่จัดเก็บชั่วคราว)** ซึ่งมีลักษณะสำคัญที่ผู้จัดการต้องเข้าใจ:
1. **ความไม่ถาวร (Ephemeral Nature):** ไฟล์ใดๆ ก็ตามที่อัปโหลดหรือดาวน์โหลดเข้ามาสู่พื้นที่เครื่องเสมือน จะคงอยู่เฉพาะขณะที่ Runtime กำลังทำงานอยู่เท่านั้น
2. **การหมดเวลาทำงาน (Timeout):** หากไม่มีการใช้งานเกินช่วงเวลาที่กำหนด (Idle Timeout เช่น 60-90 นาที) หรือใช้งานต่อเนื่องครบขีดจำกัดสูงสุด (Maximum Runtime เช่น 12 ชั่วโมง) ระบบจะทำการ Reset เครื่องเสมือนใหม่ ส่งผลให้ไฟล์และตัวแปรทั้งหมดสูญหาย
3. **แนวทางปฏิบัติที่ดี (Best Practice):** การทำงานที่มีข้อมูลสำคัญ ควรเชื่อมต่อกับ Cloud Storage ภายนอก เช่น Google Drive หรือดึงข้อมูลผ่าน API อัตโนมัติ (เช่น KaggleHub API)

````python
# นำเข้าไลบรารี kagglehub และ pandas
import kagglehub
import pandas as pd
import os

# ดาวน์โหลดเวอร์ชันล่าสุดของชุดข้อมูล gregorut/videogamesales จาก Kaggle
# คำสั่งนี้จะส่งกลับค่า path ซึ่งเป็นโฟลเดอร์ที่เก็บไฟล์ข้อมูลในเครื่องเสมือน
path = kagglehub.dataset_download("gregorut/videogamesales")

# แสดงตำแหน่งไดเรกทอรีที่ดาวน์โหลดไฟล์ลงมาเก็บไว้
print("ตำแหน่งจัดเก็บชุดข้อมูล (Dataset Path):", path)

# ตรวจสอบรายชื่อไฟล์ทั้งหมดที่อยู่ภายในไดเรกทอรีดังกล่าว
files_in_dataset = os.listdir(path)
print("รายชื่อไฟล์ในโฟลเดอร์:", files_in_dataset)

# นำเข้าไฟล์ vgsales.csv เข้าสู่ตัวแปรตารางข้อมูล df (DataFrame)
df = pd.read_csv(path + "/vgsales.csv")

# แสดงผล 5 แถวแรกของตารางข้อมูลด้วยฟังก์ชัน display()
display(df.head())
````
