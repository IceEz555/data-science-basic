## 3. ชนิดข้อมูลพื้นฐาน (Data Types) และผลกระทบต่อการวิเคราะห์ข้อมูล

จากตารางยอดขายวิดีโอเกมข้างต้น ข้อมูลแต่ละมิติจะมีชนิดข้อมูลกำกับไว้ ซึ่งส่งผลโดยตรงต่อการประมวลผล:

| ชนิดข้อมูล | ชื่อใน Python | ตัวอย่างในชุดข้อมูลเกม | บริบทการใช้งานในธุรกิจ |
|---|---|---|---|
| **ข้อความ (String)** | `str` | `"Wii Sports"`, `"Nintendo"` | ชื่อสินค้า, ผู้จัดจำหน่าย, หมวดหมู่เกม |
| **จำนวนเต็ม (Integer)** | `int` | `1`, `2006` | อันดับยอดขาย (Rank), ปีที่วางจำหน่าย (Year) |
| **ทศนิยม (Float)** | `float` | `82.74`, `41.49` | ยอดขายรวมทั่วโลก (Global_Sales), ยอดขายในทวีป |
| **ค่าความจริง (Boolean)** | `bool` | `True`, `False` | สถานะเกมระดับ Blockbuster, วางจำหน่ายในญี่ปุ่นหรือไม่ |

### กรณีศึกษาความเสี่ยง: ความผิดพลาดจากการแปลงชนิดข้อมูล (Type Mismatch Risk)
หากยอดขายถูกบันทึกเป็นข้อความ (`str`) แทนที่จะเป็นตัวเลข (`float`/`int`) เครื่องหมายบวก `+` จะกลายเป็นการต่อข้อความ ไม่ใช่การบวกเลขทางคณิตศาสตร์ เช่น:
- เมื่อคำนวณแบบข้อความ: `"41.49" + "29.02"` จะได้ผลลัพธ์เป็น `"41.4929.02"`
- เมื่อคำนวณแบบตัวเลข: `41.49 + 29.02` จะได้ผลลัพธ์ที่ถูกต้องเป็น `70.51`

โค้ดด้านล่างแสดงการกำหนดตัวแปรและการตรวจสอบชนิดข้อมูลใน Python:

```python
# การกำหนดตัวแปรพื้นฐานตามมิติของข้อมูลเกมอันดับ 1 (Wii Sports)
game_title = "Wii Sports"          # ข้อความ (str)
release_year = 2006                 # จำนวนเต็ม (int)
na_sales = 41.49                    # ทศนิยม (float)
eu_sales = 29.02                    # ทศนิยม (float)
is_blockbuster = True               # บูลีน (bool)

# การคำนวณยอดขายรวมของตลาดตะวันตก (NA + EU)
western_sales = na_sales + eu_sales

# แสดงผลลัพธ์พร้อมทั้งชนิดข้อมูลของแต่ละตัวแปรด้วยฟังก์ชัน type()
print("ชื่อเกม:", game_title, "| ชนิดข้อมูล:", type(game_title))
print("ปีที่วางจำหน่าย:", release_year, "| ชนิดข้อมูล:", type(release_year))
print("ยอดขาย NA:", na_sales, "ล้านชุด | ชนิดข้อมูล:", type(na_sales))
print("สถานะเกมระดับ Blockbuster:", is_blockbuster, "| ชนิดข้อมูล:", type(is_blockbuster))
print("ยอดขายรวมตลาดตะวันตก:", western_sales, "ล้านชุด | ชนิดข้อมูล:", type(western_sales))
```

## 4. โครงสร้างการเก็บข้อมูล: List และ Dictionary

ในการประมวลผลข้อมูลระดับองค์กร ข้อมูลไม่ได้จัดเก็บเป็นตัวแปรเดี่ยว แต่จัดเก็บในโครงสร้างข้อมูล 2 รูปแบบหลัก:

### 4.1 รายการข้อมูลตามลำดับ (List)
- สัญลักษณ์: ใช้วงเล็บก้ามปู `[ ]`
- หน้าที่: ใช้เก็บชุดข้อมูลที่มีลำดับแน่นอน เช่น ยอดขายของเกม 5 อันดับแรก
- การเข้าถึงข้อมูล: ใช้ดัชนีระบุตำแหน่ง (Index) เริ่มต้นจากตำแหน่งที่ `0`

### 4.2 คู่ค่าข้อมูลแบบมีชื่อกำกับ (Dictionary)
- สัญลักษณ์: ใช้วงเล็บปีกกา `{ }` ในรูปแบบ `{"Key": Value}`
- หน้าที่: ใช้เก็บข้อมูลระเบียน (Record) ที่มีหลายมิติ เทียบเท่ากับ **1 แถวในตารางฐานข้อมูล**
- การเข้าถึงข้อมูล: เรียกผ่านชื่อคอลัมน์ (Key Name) โดยตรง ทำให้โค้ดมีความชัดเจนในเชิงธุรกิจ

โค้ดด้านล่างแสดงการจัดเก็บข้อมูลเกมด้วย List และ Dictionary:

```python
# ตัวอย่างที่ 1: การใช้ List จัดเก็บตัวเลขยอดขายทั่วโลกของเกม 5 อันดับแรก (หน่วย: ล้านชุด)
top5_sales = [82.74, 40.24, 35.82, 33.00, 31.37]
print("ยอดขายอันดับที่ 1 (ดัชนี 0):", top5_sales[0], "ล้านชุด")
print("ยอดขายอันดับที่ 5 (ดัชนี 4):", top5_sales[4], "ล้านชุด")
print("จำนวนรายการที่บันทึก:", len(top5_sales))

print("-" * 60)

# ตัวอย่างที่ 2: การใช้ Dictionary จัดเก็บข้อมูลเกม 1 เกม (1 Record)
game_record = {
    "Rank": 1,
    "Name": "Wii Sports",
    "Platform": "Wii",
    "Year": 2006,
    "Genre": "Sports",
    "Publisher": "Nintendo",
    "Global_Sales": 82.74
}

# เข้าถึงข้อมูลโดยใช้ Key Name ที่ตรงกับมิติทางธุรกิจ
print("ชื่อเกม:", game_record["Name"])
print("แพลตฟอร์ม:", game_record["Platform"])
print("ผู้จัดจำหน่าย:", game_record["Publisher"])
print("ยอดขายทั่วโลก:", game_record["Global_Sales"], "ล้านชุด")
```

### โครงสร้างข้อมูลแบบผสม (List of Dictionaries): รากฐานของตารางข้อมูล
เมื่อนำ Dictionary หลายๆ ระเบียนมารวมกันใน List เราจะได้โครงสร้างข้อมูลที่เรียกว่า **List of Dictionaries** ซึ่งเป็นโครงสร้างมาตรฐานสากลในการแลกเปลี่ยนข้อมูลผ่าน API และเป็นจุดเริ่มต้นของการสร้างตารางใน Pandas หรือฐานข้อมูล SQL

```python
# การจัดเก็บข้อมูลเกมหลายรายการในรูปแบบ List of Dictionaries
games_dataset = [
    {"Name": "Wii Sports", "Platform": "Wii", "Genre": "Sports", "Publisher": "Nintendo", "Global_Sales": 82.74},
    {"Name": "Super Mario Bros.", "Platform": "NES", "Genre": "Platform", "Publisher": "Nintendo", "Global_Sales": 40.24},
    {"Name": "Mario Kart Wii", "Platform": "Wii", "Genre": "Racing", "Publisher": "Nintendo", "Global_Sales": 35.82},
    {"Name": "Wii Sports Resort", "Platform": "Wii", "Genre": "Sports", "Publisher": "Nintendo", "Global_Sales": 33.00},
    {"Name": "Grand Theft Auto V", "Platform": "PS3", "Genre": "Action", "Publisher": "Take-Two Interactive", "Global_Sales": 21.40}
]

print("จำนวนเกมในชุดข้อมูลจำลอง:", len(games_dataset), "เกม")
print("ข้อมูลเกมลำดับแรก:", games_dataset[0])
```

## 5. โครงสร้างเงื่อนไขสำหรับตรรกะทางธุรกิจ (Conditional Statements: if-elif-else)

โครงสร้างเงื่อนไขใช้ในการแปลง **นโยบายทางธุรกิจ (Business Rules)** ให้เป็นกฎเกณฑ์การประมวลผลของระบบคอมพิวเตอร์

- `if`: เงื่อนไขหลักข้อแรก ถ้าเงื่อนไขนี้เป็นจริง ระบบจะทำงานในบล็อกนี้และข้ามเงื่อนไขที่เหลือทันที
- `elif` (Else If): เงื่อนไขเพิ่มเติมลำดับถัดไป ในกรณีที่เงื่อนไขก่อนหน้าไม่เป็นจริง
- `else`: เงื่อนไขทางเลือกสุดท้าย สำหรับทุกกรณีที่ไม่ตรงกับเงื่อนไขข้างต้นทั้งหมด

ตรรกะสำคัญ: ระบบจะตรวจสอบจากบนลงล่างทันทีที่พบเงื่อนไขที่เป็นจริงข้อแรก ระบบจะหยุดตรวจสอบเงื่อนไขที่เหลือ

## 6. โครงสร้างการทำงานซ้ำ (Loops: for loop)

ในการทำงานจริง ข้อมูลมักมีปริมาณมากหลักร้อยถึงหลักหมื่นรายการ คำสั่ง `for loop` ทำหน้าที่ดึงข้อมูลออกมาประมวลผล **ทีละรายการตามลำดับ (Batch Iteration)**

เมื่อทำงานร่วมกันระหว่าง `for loop` และ `if-elif-else` จะเกิดกระบวนการประมวลผลข้อมูลอัตโนมัติ เช่น การจัดกลุ่มระดับความสำเร็จของเกม (Product Tiering) และคำนวณยอดขายรวม

## 7. โค้ดตัวอย่าง: การประมวลผลยอดขายและจัดกลุ่มระดับเกม (Tiering Logic)

**เกณฑ์การจัดระดับความสำเร็จของเกมโดยฝ่ายบริหาร:**
1. เกมที่มียอดขายทั่วโลก **มากกว่าหรือเท่ากับ 30 ล้านชุด** จัดเป็น `"Mega Blockbuster (Tier 1)"`
2. เกมที่มียอดขายทั่วโลก **ตั้งแต่ 20 ถึง 29.99 ล้านชุด** จัดเป็น `"Major Hit (Tier 2)"`
3. เกมที่มียอดขาย **น้อยกว่า 20 ล้านชุด** จัดเป็น `"Standard Success (Tier 3)"`

โค้ดด้านล่างแสดงการแปลงนโยบายข้างต้นเป็นคำสั่ง Python พร้อมคำอธิบายทุกบรรทัด:

```python
# ชุดข้อมูลเกมตัวอย่าง
sample_games = [
    {"Name": "Wii Sports", "Platform": "Wii", "Global_Sales": 82.74},
    {"Name": "Super Mario Bros.", "Platform": "NES", "Global_Sales": 40.24},
    {"Name": "Grand Theft Auto V", "Platform": "PS3", "Global_Sales": 21.40},
    {"Name": "Super Mario Land", "Platform": "GB", "Global_Sales": 18.14},
    {"Name": "Gran Turismo 3", "Platform": "PS2", "Global_Sales": 14.98}
]

# ตัวแปรสำหรับเก็บยอดขายรวมของเกมกลุ่มนี้
total_global_sales = 0.0

# เริ่มต้น Loop เพื่อเข้าถึงข้อมูลเกมแต่ละเกมทีละแถว
for game in sample_games:
    # ดึงค่าตัวเลขยอดขายและชื่อเกม
    title = game["Name"]
    platform = game["Platform"]
    sales = game["Global_Sales"]
    
    # ตรวจสอบเงื่อนไขตามเกณฑ์ทางธุรกิจด้วย if-elif-else
    if sales >= 30.0:
        tier = "Mega Blockbuster (Tier 1)"
    elif sales >= 20.0:
        tier = "Major Hit (Tier 2)"
    else:
        tier = "Standard Success (Tier 3)"
        
    # สะสมยอดขายรวม
    total_global_sales = total_global_sales + sales
    
    # แสดงผลการวิเคราะห์รายเกม
    print(f"เกม: {title:<25} | แพลตฟอร์ม: {platform:<6} | ยอดขาย: {sales:>6.2f} ล้านชุด | ระดับ: {tier}")

# สรุปยอดขายรวมหลังจบ Loop
print("-" * 80)
print(f"ยอดขายรวมทั้งหมด: {total_global_sales:,.2f} ล้านชุด")
```

```python
# ข้อมูลเริ่มต้นสำหรับแบบฝึกหัด
series_sales = [14.76, 14.64, 14.03, 13.73, 13.51]

# TODO: เขียนโค้ดตอบโจทย์ที่ 1
# -----------------------------------------------------------------------------
# คำนวณผลรวมยอดขายและค่าเฉลี่ยยอดขาย

total_series_sales = sum(series_sales)
average_series_sales = total_series_sales / len(series_sales)

print(f"ยอดขายรวมของซีรีส์: {total_series_sales:.2f} ล้านชุด")
print(f"ค่าเฉลี่ยยอดขายของซีรีส์: {average_series_sales:.2f} ล้านชุด")

# -----------------------------------------------------------------------------

# TODO: เขียนโค้ดตอบโจทย์ที่ 2
# -----------------------------------------------------------------------------
# ตรวจสอบเงื่อนไขตามเกณฑ์ KPI ด้วย if-elif-else

if average_series_sales >= 14.0:
    kpi_status = "สถานะ: ผลงานระดับยอดเยี่ยมบรรลุเป้าหมายสูงสุด"
elif 12 < average_series_sales <= 13.99:
    kpi_status = "สถานะ: ผลงานระดับมาตรฐานตามเป้าหมาย"
else:
    kpi_status = "สถานะ: ยอดขายต่ำกว่าเกณฑ์ ต้องทบทวนแผนพัฒนาสินค้า"

print(kpi_status)

# -----------------------------------------------------------------------------
```

## เฉลยและคำอธิบายตรรกะคำสั่ง (Solution & Logical Explanation)

ส่วนนี้จัดทำขึ้นเพื่อให้ผู้บริหารสามารถตรวจสอบวิธีคิดและคำตอบที่ถูกต้อง

```python
# เฉลยโจทย์ที่ 1: การคำนวณผลรวมและค่าเฉลี่ย
series_sales = [14.76, 14.64, 14.03, 13.73, 13.51]

# คำนวณผลรวมโดยใช้ฟังก์ชัน sum() ในตัวของ Python
total_series_sales = sum(series_sales)

# คำนวณจำนวนข้อมูลทั้งหมดโดยใช้ฟังก์ชัน len()
game_count = len(series_sales)

# คำนวณค่าเฉลี่ยโดยนำผลรวมหารด้วยจำนวนเกม
avg_series_sales = total_series_sales / game_count

print("ผลรวมยอดขายทั้งหมด:", f"{total_series_sales:,.2f}", "ล้านชุด")
print("ค่าเฉลี่ยยอดขายต่อภาค:", f"{avg_series_sales:,.2f}", "ล้านชุด")

# เฉลยโจทย์ที่ 2: การประเมินสถานะ KPI ด้วยโครงสร้าง if-elif-else
if avg_series_sales >= 14.0:
    kpi_status = "สถานะ: ผลงานระดับยอดเยี่ยม บรรลุเป้าหมายสูงสุด"
elif avg_series_sales >= 12.0:
    kpi_status = "สถานะ: ผลงานระดับมาตรฐานตามเป้าหมาย"
else:
    kpi_status = "สถานะ: ยอดขายต่ำกว่าเกณฑ์ ต้องทบทวนแผนพัฒนาสินค้า"

print(kpi_status)
```

### คำอธิบายเชิงตรรกะสำหรับผู้บริหาร
1. **การคำนวณเชิงคณิตศาสตร์ (`sum()` และ `len()`):**
   - `sum(series_sales)` รวมตัวเลขทั้งหมดใน List ได้อย่างรวดเร็ว ได้ผลลัพธ์เป็น 70.67 ล้านชุด
   - `len(series_sales)` คืนค่าจำนวนเกม (5 ภาค) ทำให้ค่าเฉลี่ยออกมาเป็น 14.13 ล้านชุด
2. **การจัดลำดับเงื่อนไข (`if-elif-else`):**
   - เนื่องจากค่าเฉลี่ยอยู่ที่ 14.13 ล้านชุด ซึ่งมากกว่า 14.0 ระบบจะเข้าเงื่อนไขแรกทันทีและแสดงผลว่า "ผลงานระดับยอดเยี่ยม บรรลุเป้าหมายสูงสุด"
   - การเขียนเงื่อนไขโดยเรียงจากค่าสูงสุดลงมาค่าต่ำสุด ทำให้ไม่เกิดข้อผิดพลาดในการดักจับเงื่อนไข
