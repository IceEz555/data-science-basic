# 📊　Dataset & Libraries Used

```python
# IMPORTING DATASET #
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
plt.rcParams['figure.dpi'] = 120

golf_dataset = pd.read_csv("Data/golf_dataset_mini_original_with_testset.csv")
df = golf_dataset.copy()

df
```

# Algorithm Steps

# 　Preparation Phase
![Imgur](https://i.imgur.com/1sPf4Rw.png)

```python
from sklearn.model_selection import train_test_split

# Set feature matrix X and target vector y
X, y = df.drop('Play', axis=1), df['Play']

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.5, shuffle=False)
```

![Imgur](https://i.imgur.com/adP2POv.png)

```python
from sklearn.dummy import DummyClassifier

# Choose a strategy for your DummyClassifier (e.g., 'most_frequent', 'stratified', etc.)
strategy = 'most_frequent'
```

# 　Training Phase
![Imgur](https://i.imgur.com/O24Sm7r.png)

```python
# Initialize the DummyClassifier
dummy_clf = DummyClassifier(strategy=strategy)

# "Train" the DummyClassifier (although no real training happens)
dummy_clf.fit(X_train, y_train)
```

# 　Classification Phase
![Imgur](https://i.imgur.com/68hfsww.png)

```python
# Use the DummyClassifier to make predictions
y_pred = dummy_clf.predict(X_test)
print("Label     :",list(y_test))
print("Prediction:",list(y_pred))
```

# 　Evaluation Phase
![Imgur](https://i.imgur.com/aq6p3ol.png)

```python
# Evaluate the DummyClassifier's accuracy
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)
print(f"Dummy Classifier Accuracy: {accuracy.round(4)*100}%")
```

# Key Parameters
![](https://i.imgur.com/8hSJ30o.png)

```python
labels, predictions, accuracies = list(y_test), [], []

strategies = ['stratified', 'most_frequent', 'uniform']
for strategy in strategies:
    dummy_clf = DummyClassifier(strategy=strategy, random_state=42)
    dummy_clf.fit(X_train, y_train)
    y_pred = dummy_clf.predict(X_test)
    predictions.append(list(y_pred))
    accuracies.append(accuracy_score(y_test, y_pred).round(4)*100)

df_predictions = pd.DataFrame({'Label': labels})
for strategy, pred in zip(strategies, predictions):
    df_predictions[f'{strategy}'] = pred

df_accuracies = pd.DataFrame({'Accuracy ': accuracies}, index=[f'{strategy}' for strategy in strategies]).T

print(df_predictions)
print(df_accuracies)
```

# 🌟 Dummy Classifier Simplified

```python
from sklearn.model_selection import train_test_split
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score

X, y = df.drop('Play', axis=1), df['Play']
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.5, shuffle=False)

dummy_clf = DummyClassifier(strategy='most_frequent')
dummy_clf.fit(X_train, y_train)
y_pred = dummy_clf.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, y_pred)}")
```
